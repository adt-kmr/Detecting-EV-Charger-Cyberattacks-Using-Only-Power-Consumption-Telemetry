# 02 â€” Chronological Split Method**Goal:** Evaluate binary and multiclass cyberattack detection under a *strict chronological* (out-of-distribution) train/test split that avoids temporal leakage.**Key idea:** Train on the first 80% of the timeline, test on the final 20% (future, unseen). Real-world power telemetry has strong temporal structure, so this gives an honest measure of forward generalization.**Approach:** Chronological split, rich rolling features (`delta`, `mean`, `std`, `skew`, `crest`), Isolation-Forest anomaly score, **soft-voting ensemble** (XGBoost + LightGBM + RandomForest), multiclass XGBoost on `Attack-Group`, SHAP.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import IsolationForest, RandomForestClassifier, VotingClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    precision_recall_curve,
    auc,
    balanced_accuracy_score
)
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
import shap

# Fixes the Matplotlib bug natively and applies a clean, publication-grade theme
sns.set_theme(style="whitegrid", context="notebook", palette="muted")
plt.rcParams["figure.figsize"] = (8, 5)
warnings.filterwarnings('ignore')
```

```python
# Load dataset from Colab environment
df = pd.read_csv("EVSE-B-PowerCombined (1).csv")
print(f"Original Shape: {df.shape}")

# 1. TEMPORAL INTEGRITY CHECK
df['time'] = pd.to_datetime(df['time'])
df = df.sort_values(by='time').reset_index(drop=True)
print(f"Chronological Timeline Cover: {df['time'].min()} to {df['time'].max()}")

# Base physical features
signals = ["shunt_voltage", "bus_voltage_V", "current_mA", "power_mW"]
window_size = 20

# 2. HIGH-ORDER TRANSIENT ENGINE
print("Engineering high-order statistical and differential transient features...")
for col in signals:
    # Captures localized rate-of-change spikes
    df[f"{col}_delta"] = df[col].diff()

    # Captures localized structural shifts
    df[f"{col}_mean"] = df[col].rolling(window=window_size).mean()
    df[f"{col}_std"] = df[col].rolling(window=window_size).std()

    # Novel Addition: Skewness captures asymmetric wave adjustments caused by line injection attacks
    df[f"{col}_skew"] = df[col].rolling(window=window_size).skew()

    # Crest factor approximation (Peak-to-Average ratio indicating signature distortions)
    df[f"{col}_crest"] = df[col] / (df[f"{col}_mean"] + 1e-5)

# Drop rows with NaN values resulting from looking backward during rolling window operations
df = df.dropna().reset_index(drop=True)
print(f"Shape post-feature engineering: {df.shape}")
```

```python
print("Computing Unsupervised Spatial Behavioral Index using Isolation Forest...")

# Train an Unsupervised Isolation Forest on pure physical signals
# to calculate structural anomalies independent of the labels
iso_forest = IsolationForest(contamination=0.10, random_state=42, n_jobs=-1)

# CRITICAL FIX: Fit the model on the background measurements first
iso_forest.fit(df[signals])

# .decision_function returns an anomaly score (lower means more anomalous)
df['unsupervised_anomaly_score'] = iso_forest.decision_function(df[signals])

print("Sample anomaly scores integrated successfully.")
print(df[['unsupervised_anomaly_score', 'Label']].head())
```

```python
# Separate out all target columns and non-numeric metadata
metadata_cols = ["time", "State", "Attack", "Attack-Group", "Label", "interface"]
feature_cols = [c for c in df.columns if c not in metadata_cols]

X = df[feature_cols]

# Target 1: Binary Classification Target
le_binary = LabelEncoder()
y_binary = le_binary.fit_transform(df["Label"])

# CRITICAL RESEARCH PROTECTION: Chronological Split (Prevents 99.9% artificial leakage)
# Train on the first 80% chronologically, validate on the remaining 20% future stream
split_index = int(len(df) * 0.80)

X_train, X_test = X.iloc[:split_index], X.iloc[split_index:]
y_train_bin, y_test_bin = y_binary[:split_index], y_binary[split_index:]

print(f"Leakage-Free Train Matrix Shape: {X_train.shape}")
print(f"Leakage-Free Test Matrix Shape: {X_test.shape}")
print(f"Test Class Balance Ratio:\n {pd.Series(y_test_bin).value_counts(normalize=True)}")
```

```python
# Construct diverse sub-classifiers configured to handle imbalanced tracking data
xgb_model = XGBClassifier(n_estimators=100, max_depth=5, learning_rate=0.05, eval_metric="logloss", random_state=42)
lgb_model = LGBMClassifier(n_estimators=100, max_depth=5, learning_rate=0.05, random_state=42, verbose=-1)
rf_model = RandomForestClassifier(n_estimators=100, max_depth=8, random_state=42, n_jobs=-1)

# Combine into a heterogeneous Soft Voting Ensemble
hybrid_ensemble = VotingClassifier(
    estimators=[('xgb', xgb_model), ('lgb', lgb_model), ('rf', rf_model)],
    voting='soft'
)

print("Training Advanced Hybrid Stacking Ensemble across chronological timeline...")
hybrid_ensemble.fit(X_train, y_train_bin)

# Predictions and Evaluative Metrics
binary_preds = hybrid_ensemble.predict(X_test)
binary_probs = hybrid_ensemble.predict_proba(X_test)[:, 1]

print("\n" + "="*20 + " HYBRID ENSEMBLE RESULTS " + "="*20)
print(f"Balanced Accuracy Score: {balanced_accuracy_score(y_test_bin, binary_preds):.4f}")
print("\nDetailed Publication Report:\n", classification_report(y_test_bin, binary_preds, target_names=le_binary.classes_))

# PR-AUC Calculation (Essential for validation metrics in high-fidelity security papers)
precision, recall, _ = precision_recall_curve(y_test_bin, binary_probs)
pr_auc = auc(recall, precision)
roc_auc = roc_auc_score(y_test_bin, binary_probs)
print(f"Authentic ROC-AUC: {roc_auc:.4f} | Robust Precision-Recall AUC: {pr_auc:.4f}")

# Plot Confusion Matrix
cm = confusion_matrix(y_test_bin, binary_preds)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=le_binary.classes_, yticklabels=le_binary.classes_)
plt.title("Ensemble Confusion Matrix (Chronological Evaluation)")
plt.ylabel("Ground Truth")
plt.xlabel("Predicted Category")
plt.tight_layout()
plt.show()
```

```python
# Target 2: Deep mapping into specific attack groups (DoS, Recon, Host-Attack)
le_multi = LabelEncoder()
y_multiclass = le_multi.fit_transform(df["Attack-Group"])

y_train_mul, y_test_mul = y_multiclass[:split_index], y_multiclass[split_index:]

# Deploy Multiclass variant of XGBoost Engine
xgb_multi = XGBClassifier(
    n_estimators=120,
    max_depth=5,
    learning_rate=0.05,
    objective="multi:softprob",
    eval_metric="mlogloss",
    random_state=42
)

print("Training Multi-Class Threat Classifier...")
xgb_multi.fit(X_train, y_train_mul)
multi_preds = xgb_multi.predict(X_test)

print("\n" + "="*20 + " MULTICLASS EXTENSION RESULTS " + "="*20)
print(f"Balanced Accuracy Score: {balanced_accuracy_score(y_test_mul, multi_preds):.4f}")
print("\nClassification Vector Matrix:\n", classification_report(y_test_mul, multi_preds, target_names=le_multi.classes_))

# Multi-Class Heatmap Visual
cm_multi = confusion_matrix(y_test_mul, multi_preds)
plt.figure(figsize=(7, 5))
sns.heatmap(cm_multi, annot=True, fmt="d", cmap="YlOrRd", xticklabels=le_multi.classes_, yticklabels=le_multi.classes_)
plt.title("Multiclass Cyberattack Signature Mapping")
plt.ylabel("True Attack Group Vector")
plt.xlabel("Predicted Attack Group Vector")
plt.tight_layout()
plt.show()
```

```python
print("Initializing SHAP explainability mapping (sampling test subset for computational speed)...")

# Sample a random snapshot of the unseen test set to run feature space impacts
test_sample = X_test.sample(n=min(400, len(X_test)), random_state=42)

# Extract underlying trained XGBoost model from the binary ensemble for SHAP interpretation
trained_xgb = hybrid_ensemble.named_estimators_['xgb']

explainer = shap.TreeExplainer(trained_xgb)
shap_values = explainer.shap_values(test_sample)

# Plot 1: SHAP Global Feature Importance
plt.figure(figsize=(10, 5))
shap.summary_plot(shap_values, test_sample, plot_type="bar", show=False)
plt.title("SHAP Global Impact Profile: How Engineered Micro-Transients Drive Decisions", pad=15)
plt.tight_layout()
plt.show()

# Plot 2: SHAP Beeswarm Plot (Shows directional impact of high vs low signal fluctuations)
plt.figure(figsize=(10, 5))
shap.summary_plot(shap_values, test_sample, show=False)
plt.title("SHAP Feature Dynamics: Impact Direction of Electrical Fluctuations", pad=20)
plt.tight_layout()
plt.show()
```
